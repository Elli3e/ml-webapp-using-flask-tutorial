```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
data_sh = pd.read_csv("C:/Users/e_bab/Documents/4geeks/python-hello/ml-webapp-using-flask-tutorial/Youtube05-Shakira.csv")
data_m= pd.read_csv("C:/Users/e_bab/Documents/4geeks/python-hello/ml-webapp-using-flask-tutorial/Youtube04-Eminem.csv")
data_lm=pd.read_csv("C:/Users/e_bab/Documents/4geeks/python-hello/ml-webapp-using-flask-tutorial/Youtube03-LMFAO.csv")
data_kp=pd.read_csv("C:/Users/e_bab/Documents/4geeks/python-hello/ml-webapp-using-flask-tutorial/Youtube02-KatyPerry.csv")
data_ps=pd.read_csv("C:/Users/e_bab/Documents/4geeks/python-hello/ml-webapp-using-flask-tutorial/Youtube01-Psy..csv")
print(data_sh.head(2))
print(data_sh.shape)
print(data_m.head(2))
print(data_m.shape)
print(data_lm.head(2))
print(data_lm.shape)
print(data_kp.head(2))
print(data_kp.shape)
print(data_ps.head(2))
print(data_ps.shape)
```

```python
data_tot= pd.concat([data_sh,data_m, data_lm,data_kp,data_ps], axis =0, ignore_index=True)
print(data_tot.head(3))
print(data_tot.shape)
```

```python
print(data_tot.info())
print(data_tot.isnull().sum())
print(data_tot.duplicated().sum())
```

```python
print(data_tot["CONTENT"].duplicated().sum())
content_duplicate= data_tot[data_tot["CONTENT"].duplicated(keep=False)]
print(content_duplicate[["CONTENT", "CLASS"]].sort_values("CONTENT"))
```

```python
print("Total rows:", len(data_tot))
print("Unique comment IDs:", data_tot["COMMENT_ID"].nunique())
print("Duplicate comment IDs:", data_tot["COMMENT_ID"].duplicated().sum())
data_tot.drop_duplicates()
```

```python
data_tot["CONTENT"]=data_tot["CONTENT"].str.strip()
data_tot["CONTENT"]=data_tot["CONTENT"].str.lower()
print(data_tot["CONTENT"].str.contains(r"[?,_\-,!]").sum())
```

```python
from sklearn.model_selection import train_test_split

x = data_tot["CONTENT"]
y= data_tot["CLASS"]

x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2, random_state=42)
print(x_train.head(2))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()

x_train_DF= vectorizer.fit_transform(x_train)
x_test_DF= vectorizer.transform(x_test)
print(x_train_DF[:2])
```

```python
from sklearn.svm import SVC
model_SVC= SVC(kernel="linear")
model_SVC.fit(x_train_DF, y_train)

y_pred=model_SVC.predict(x_test_DF)

print(y_pred)

from sklearn.metrics import classification_report
print(f"accuracy score is : {classification_report(y_test, y_pred)}")
```

```python
# comparing with keras dense, neural network 
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Dropout

tf_model = Sequential([Dense(128, activation = "relu"),
                       Dropout(0.3),
                       Dense(64, activation = "relu"), 
                       Dropout(0.3),
                       Dense(1, activation="sigmoid")
])
```

```python
from tensorflow.keras.optimizers import Adam

tf_model.compile(optimizer=Adam(learning_rate = 0.001), loss="binary_crossentropy", metrics=["accuracy",tf.keras.metrics.Precision(),
    tf.keras.metrics.Recall()])
```

```python
tf_model.fit(x_train_DF, y_train)
y_pred_prob= tf_model.predict(x_test_DF)
print(y_pred_prob)

y_pred= (y_pred_prob>0.5).astype(int).reshape(-1,)
print(classification_report(y_test, y_pred))
```

```python
#using LSTM 
tf_vectorizer = tf.keras.layers.TextVectorization(
    max_tokens=10000,
    output_mode="int",
    output_sequence_length=50
)
```

```python
tf_vectorizer.adapt(x_train)
x_train_seq = tf_vectorizer(x_train)
x_test_seq = tf_vectorizer(x_test)
print(x_train_seq[:2])
print(x_train_seq.shape)
```

```python
print(x_train_seq.shape)
print(x_test_seq.shape)

print(x_train_seq[:5])
print(x_test_seq[:5])
```

```python
vocab_size = len(tf_vectorizer.get_vocabulary())

print(vocab_size)
```

```python
from tensorflow.keras.layers import Embedding 
from tensorflow.keras.layers import LSTM

LSTM_model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=64
    ),
    LSTM(64),
    Dense(1, activation="sigmoid")
])
```

```python
LSTM_model.compile(optimizer= Adam(learning_rate =0.001), loss="binary_crossentropy",metrics=["accuracy"])
```

```python
LSTM_model.fit(x_train_seq, y_train, epochs = 10, batch_size = 32)
y_pred_prob= LSTM_model.predict(x_test_seq)
print(y_pred_prob)

# 
```

```python
y_pred= (y_pred_prob>0.5).astype(int).reshape(-1,)
print(classification_report(y_test, y_pred))
```
